```python
def is_prime(num:int)->bool:


    '''
    this function check my number is prime number or not ( T or F)
    Args:
        num(int): input number
    returns:
        bool
                
    '''
    IsPrime=True
    if num<=1:
        IsPrime=False
    else:
        for i in range(2,int(num ** 0.5)+1):
            if num % i ==0:
                IsPrime=False
                break   
 
    return IsPrime

print(is_prime(17))
```

```python
def factorial(num:int)->int:
    if num==0:
        return 1
    return num*factorial(num-1)
print(factorial(500))
```

```python
def common_divisors(num1:int , num2:int)-> list[int]:
    limit= min(num1,num2)
    divisor=[]
    for i in range(1,limit+1):
        if num1 % i == 0 and num2 % i ==0:
            divisor.append(i)
    return divisor        


common_divisors(10,20)
```

```python
def reverse_words(text:str)->str:
    lst= text.split()
    new_list=lst[::-1]
    new_str=' '.join(new_list)
    return new_str


print(reverse_words("omar mohmaed"))
```

```python
def dect(words:list)->dict:
    dect_words={}
    unique_words = set(words)
    for word in unique_words:
        dect_words.update({word:words.count(word)})
    return dect_words    
    

        
words = ["Welcome", "Ali", "Hi", "Ali", "No", "Hi", "No", "Ali", "No", "Ali"]
print(dect(words))

```

```python
responses = {
    "hello": ["Hello!", "Hi there!", "Greetings!"],
    "how are you": ["I'm doing well, thank you!", "I'm fine, how about you?"],
    "goodbye": ["Goodbye!", "See you later!", "Farewell!"],
    "default": ["I'm sorry, I didn't understand.", "Could you please rephrase that?"]
}
 
def get_responses(user_input):
    for key in responses:
        if key in user_input:
            return random.choice(responses[key])
    return random.choice(responses["default"])


def main():
    print("chatbot : hi how i can help you! ")
    while True:
        user_input = input("User :  ").lower()
        responses= get_responses(user_input)
        print("chatbot",responses)
        if user_input=="goodbye":
         break



main()
    
```
