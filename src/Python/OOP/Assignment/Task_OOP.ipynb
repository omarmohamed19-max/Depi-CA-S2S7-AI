```python

# Problem Number one

class MultiplicationTable:
    """A class to generate and display multiplication tables for a given number.

    Attributes:
        num (int | float): The base number for which the multiplication table is generated.
    """

    def __init__(self,num:int|float)->None:
        """Initializes the MultiplicationTable instance with a base number.

        Args:
            num (int | float): The number to generate the multiplication table for.
        """
        self.num=num

    def generate_table(self,limit:int=10)->None:
        """Prints the multiplication table up to a specified limit.

        Args:
            limit (int, optional): The upper bound of the multiplication table. Defaults to 10.
        """
        for i in range(1,limit+1):
            print(f"{self.num} * {i} = {self.num*i}")




if __name__=="__main__":
    n1=MultiplicationTable(9)
    n1.generate_table(12)

        
```

```python

# problem number two

class TwinPrime:
    """A utility class for identifying and displaying twin prime numbers within a given range."""

    def __init__(self)->None:
        pass

    @staticmethod
    def is_prime(num:int)->bool:
        """Checks whether a given integer is a prime number.

        Args:
            num (int): The integer number to be checked.

        Returns:
            bool: True if the number is prime, False otherwise.
        """

        if num<2:
                return False
        for i in range(2, int(num**0.5)+1):
                if num%i==0:
                    return False
        return True

    def display_twin_primes(self,limit:int=1000)->None:

        """Prints all twin prime pairs up to a specified upper limit.

        Args:
            limit (int, optional): The maximum limit for finding twin primes. Defaults to 1000.
        """
    
        print("Twin Prime :")
        for i in range(3,limit-1):
            if self.is_prime(i)and self.is_prime(i+2):
                print(f"{i} and {i+2}")



if __name__=="__main__":
     ob=TwinPrime()
     ob.display_twin_primes(700)
     
     

         
```

```python

# problem number three

class PrimeFactors:
    """A class to compute and retrieve the prime factors of a given integer.

    Attributes:
        num (int): The original integer number to factorize.
        factors (list[int]): Stores the computed prime factors.
    """

    def __init__(self,num:int)->None:
        """Initializes the PrimeFactors instance with a target number.

        Args:
            num (int): The integer number to find prime factors for.
        """
        self.num=num
        self.factors:list[int]=[]

    def git_prime_factors(self)->list[int]:
        """Calculates and returns the prime factors of the initialized number.

        Returns:
            list[int]: A list containing all prime factors in ascending order.
        """
        divisor=2
        temp_num=self.num
        self.factors=[]
        while divisor*divisor<=temp_num:
                while temp_num%divisor==0:
                    self.factors.append(divisor)
                    temp_num//=divisor
                divisor+=1
        if temp_num>1:
             self.factors.append(temp_num)     
        return self.factors


if __name__=="__main__":
    f1=PrimeFactors(56)
    print(f1.git_prime_factors())
        


```

```python

# problem number four

class DecToBin:
    """A class to convert a decimal (base-10) integer into its binary (base-2) representation.

    Attributes:
        num (int): The decimal integer to be converted.
    """

    def __init__(self,num:int)->None:
        """Initializes the DecToBin instance with a decimal number.

        Args:
            num (int): The non-negative decimal integer for conversion.
        """
        self.num=num

    def get_bin_num(self)->str:
            """Converts the initialized decimal number to a binary string representation.

        Returns:
            str: A string representing the binary equivalent of the decimal number.
        """

            binary=''
            temp_num=self.num

            if temp_num==1:
                 return "1"
            elif temp_num==0:
                 return "0"
            
            while temp_num>=2:
                if temp_num%2!=0:
                    binary+="1"
                    temp_num//=2
                else:
                    binary+="0"  
                    temp_num//=2
        
            binary+="1"
            binary=binary[::-1]
            return binary


if __name__=="__main__":
    n1=DecToBin(11)
    print(n1.get_bin_num())    




```

```python

# problem number five

class PerfectNums:
    """A class to find all perfect numbers within a specified range [start, end].

    Attributes:
        start (int): The starting boundary of the search range.
        end (int): The ending boundary of the search range.
    """

    def __init__(self,start:int=1,end:int=100)->None:
        """Initializes the PerfectNums instance with a range.

        Args:
            start (int, optional): Lower bound of the range. Defaults to 1.
            end (int, optional): Upper bound of the range. Defaults to 100.
        """
        self.start=max(1,start)
        self.end=end

    def get_perfect_nums(self)->list[int]:
            """Finds and returns all perfect numbers within the defined range.

        A number is perfect if it is equal to the sum of its proper positive divisors.

        Returns:
            list[int]: A list of integers representing perfect numbers found in the range.
        """
            divisors_sum=0
            perfect_list: list[int]=[]
            for i in range(self.start,self.end+1):
                for j in range(1,i):
                    if i%j==0:
                        divisors_sum+=j
                if i==divisors_sum:
                    perfect_list.append(i)
                divisors_sum=0
        
            return perfect_list

if __name__=="__main__":
    n1=PerfectNums(5,500)
    print(n1.get_perfect_nums())    


            
```
