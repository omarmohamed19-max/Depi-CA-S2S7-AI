```python
from abc import ABC,abstractmethod
class Animal(ABC):
    @abstractmethod
    def move(self):
        pass

class Bird(Animal):
    def move(self):
        print("move from Bird")
class cat(Animal):  
    def move(self):
        print("move from cat")  
a=Bird()          
a.move()          
```

```python
class rectangle:
    _length=5         #protected
    __width=10       #pravit
    def area(self):
        return self.length*self.width
r=rectangle()    
print(r._rectangle__width)
print(r._length)
print(r.area())
```

```python
class Rectangle:
    def __init__(self , length , width):
        self.__length=length
        self.width=width
    @property
    def length(self):
        return self.__length  

    @length.setter
    def length(self,value):
        if value <= 0:
            raise ValueError("length must be positive")  
        self.__length=value
```

```python
r=Rectangle(10,20)
r._length=20
```

```python
class shape:
    def calculate_area(self):
        pass

class circle(shape):
    def __init__(self,radius):
        self.radius=radius
    def calculate_area(self):
        return 3.14*(self.radius)**2

class square (shape):
    def __init__(self,side):
        self.side=side        

    def calculate_area(self):
        return self.side**2

x=circle(5)
y=square(4)
print("area : "x.calculate.area())        
print("area : "y.calculate.area())        
```

```python
class model:
    def fit(self,x,y):
        raise NotImplementedError
    def predict(self,x):
        raise NotImplementedError

class LinearRegression(model):
    def fit(self,x,y):
        print("fitting linear_regression")

    def predict(self,x):
        print("predicting with linear regression")        


class DicisionTree(model):
    def fit(self,x,y):
        print("fitting DicisionTree")

    def predict(self,x):
        print("predicting with DecisionTree")  


models=[LinearRegression(),DicisionTree()]

for model in models:
    model.fit(None,None)
    model.predict(None)
```
