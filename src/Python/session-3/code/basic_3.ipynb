```python
def train_model(model_name, *datasets , **hyperparameters):
    print(f"training model:{model_name}")
     
    print("\nDatasets")
    for data in datasets:
        print("-",data)

    print("\nhyperparameters")
    for key , value in hyperparameters.items():
        print(f"{key},{value}")     


data_list=[]
input_user='yes'
while input_user.lower()=="yes":
    data=input("enter dataset : ")  
    data_list.append(data)
    input_user=input("Do you want enter new DataSet (Yes or No): ")

   



train_model("linear_regression",*data_list,learning_rate=0.001,epochs=1000,optimizer='x')        
```
