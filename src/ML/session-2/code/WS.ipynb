```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error , r2_score , root_mean_squared_error
import warnings 
warnings.filterwarnings('ignore')
```

```python
df=pd.read_csv("Salary_Data.csv")
df.head()
print(df.shape)
```

```python
plt.figure(figsize=(3,3))
sns.pairplot(data=df, x_vars=['YearsExperience'], y_vars=['Salary'], size = 3, kind="scatter")
plt.title("salary vs years")
plt.xlabel("years")
plt.ylabel("salary")
plt.show()
```

```python
x=df.iloc[:,:-1]
y=df.iloc[:,1]
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
reg_line=LinearRegression()
reg_line.fit(x_train,y_train)

y_hat_train=reg_line.predict(x_train)
y_hat_test=reg_line.predict(x_test)

```

```python
plt.scatter(x_train,y_train,color="red")
plt.plot(x_train,y_hat_train ,color="blue")
plt.title("Salary vs experience (training set)")
plt.xlabel("years of experience")
plt.ylabel("salary")
plt.show()
```

```python
#calculat maan squred error
mse_train=mean_squared_error(y_train,y_hat_train)
mse_test=mean_squared_error(y_test,y_hat_test)
print(mse_train)
print(mse_test)

rmse=root_mean_squared_error(y_test,y_hat_test)
print(rmse)

r2_train=r2_score(y_train,y_hat_train)
r2_test=r2_score(y_test,y_hat_test)
print(r2_train)
print(r2_test)
```

```python
w=reg_line.coef_
```
